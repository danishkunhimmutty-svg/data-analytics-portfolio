```python
import numpy as np
import pandas as pd
bnk=pd.read_html('https://www.fdic.gov/bank-failures/failed-bank-list')
```

```python
bnk
```

```python
bnk=bnk[0]
bnk
```

```python
bnk.dtypes
```

```python
bnk.columns=(bnk.columns.str.lower().str.replace(' ',"_"))
bnk.columns
```

```python
bnk['closing_date']=pd.to_datetime(bnk['closing_date'])
bnk['closing_date'].head(10)
```

```python
bnk.isna().sum()
```

```python
total_failed=len(bnk)
total_failed
```

```python
bnk['closing_date'].min()
```

```python
bnk['closing_date'].max()
```

```python
state_counts=bnk['state'].value_counts()
state_counts
```

```python
state_counts.head(5)
```

```python
bnk['year']=bnk['closing_date'].dt.year
bnk['year'].head()
```

```python
failures_per_year=bnk.groupby('year').size()
failures_per_year
```

```python
bnk['year'].min()
```

```python
failures_per_year.idxmax(),failures_per_year.max()
```

```python
failures_per_year.idxmin(),failures_per_year.min()
```

```python
min_val=failures_per_year.min()
failures_per_year[failures_per_year== min_val]
```

```python
state_year_pivot=pd.pivot_table(
    bnk,
    values='bank_name',
    index='year',
    columns='state',
    aggfunc='count'
)
state_year_pivot.head()
```

```python
state_year_pivot.isna().sum()
```
